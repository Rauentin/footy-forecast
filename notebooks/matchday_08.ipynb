# Matchday 8 — 2026/27
Dixon-Coles, xi = 0.0018, results up to 19 Oct 2026.

```python
import sys
sys.path.insert(0, "../src")

import pandas as pd
from pl_model.model import ModelParams, predict_match
from pl_model.simulate import load_fixtures

params = ModelParams.load("../models/params.json")
fixtures = load_fixtures("../data/epl_2026_27_fixtures.csv")

# the fixture list is ordered by matchday, 10 games each
MATCHDAY = 8
games = fixtures[(MATCHDAY - 1) * 10 : MATCHDAY * 10]
```

```python
rows = []
for home, away in games:
    p = predict_match(params, home, away)
    outcomes = {f"{home} win": p["home_win"], "Draw": p["draw"], f"{away} win": p["away_win"]}
    pick = max(outcomes, key=outcomes.get)

    rows.append({
        "Home": home,
        "Away": away,
        "Prediction": pick,
        "Confidence": outcomes[pick],
        "Home win": p["home_win"],
        "Draw": p["draw"],
        "Away win": p["away_win"],
        "xG home": p["expected_home_goals"],
        "xG away": p["expected_away_goals"],
    })

predictions = pd.DataFrame(rows)
percent = ["Confidence", "Home win", "Draw", "Away win"]
predictions.style.format("{:.0%}", subset=percent).format("{:.2f}", subset=["xG home", "xG away"]).hide(axis="index")
```
